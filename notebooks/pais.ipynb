# Preparo

## Bibliotecas

```python
import os
import sys

import pandas as pd
import plotly.express as px
```

```python
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
```

```python
from utils.add_country_name import add_country_name
from utils.adjust_cuisines import adjust_cuisines
from utils.clean_data import clean_data
from utils.convert_to_usd import convert_to_usd
from utils.create_color_name import create_color_name
from utils.create_price_tye import create_price_tye
from utils.create_unique_restaurant_name import create_unique_restaurant_name
from utils.rename_columns import rename_columns
```

## Baixando o Dataset

```python
file_path = os.path.join("..", "database", "zomato.csv")

print(f"Tentando ler o arquivo em: {os.path.abspath(file_path)}")

try:
    df = pd.read_csv(file_path, 
                    encoding="utf-8", 
                    on_bad_lines="skip")
    print("Sucesso! O DataFrame foi carregado.")
except Exception as e:
    print(f"Erro: {e}")
```

## Limpando o Dataset

```python
df = (df.pipe(rename_columns)
                .pipe(add_country_name)
                .pipe(clean_data)
                .pipe(convert_to_usd)
                .pipe(create_price_tye)
                .pipe(create_color_name)
                .pipe(adjust_cuisines)
                .pipe(create_unique_restaurant_name))
```

# Pais

## Top 10 Quantidade de Restaurantes Registrados por País

```python
# 1. Agrupando por país e contando a quantidade de restaurantes
restaurantes_por_pais = (
    df.groupby("country")
    .size() # Conta quantas linhas (restaurantes) existem para cada país
    .reset_index(name="quantidade_restaurantes")
    .sort_values(by="quantidade_restaurantes", ascending=False)
)

# 2. Pegando o Top 10
top_7_paises = restaurantes_por_pais.head(10)

# 3. Criando o gráfico
fig = px.bar(
    top_7_paises, 
    x="country", 
    y="quantidade_restaurantes", 
    title="Top 7 Quantidade de Restaurantes Registrados por País",
    labels={
        "country": "País", 
        "quantidade_restaurantes": "Quantidade de Restaurantes"
    },
    text_auto=True,
    color="country", # Cada país terá sua própria cor
    color_discrete_sequence=px.colors.qualitative.Safe # Paleta segura
)

# 4. Ajustando o visual
fig.update_layout(
    xaxis_tickangle=-45, 
    title_x=0.5,
    showlegend=False, # Como o nome do país já está no eixo X, podemos esconder a legenda
    xaxis_categoryorder='total descending' # Força a ordem da maior barra para a menor
)

# 5. Exibindo o gráfico
fig.show()
```

## Top 10 Quantidade de Cidades Registrados por País

```python
# 1. Agrupando por país e contando as cidades únicas
cidades_por_pais = (
    df.groupby("country")["city"]
    .nunique() # Conta quantas cidades diferentes existem por país
    .reset_index(name="quantidade_cidades")
    .sort_values(by="quantidade_cidades", ascending=False)
)

# 2. Pegando o Top 10
top_10_paises_cidades = cidades_por_pais.head(10)

# 3. Criando o gráfico
fig = px.bar(
    top_10_paises_cidades, 
    x="country", 
    y="quantidade_cidades", 
    title="Top 10 Países com Mais Cidades Registradas",
    labels={
        "country": "País", 
        "quantidade_cidades": "Quantidade de Cidades"
    },
    text_auto=True,
    color="country", 
    color_discrete_sequence=px.colors.qualitative.Safe 
)

# 4. Ajustando o visual
fig.update_layout(
    xaxis_tickangle=-45, 
    title_x=0.5,
    showlegend=False, # Legenda oculta pois os países já estão no eixo X
    xaxis_categoryorder='total descending' # Força a ordem decrescente
)

# 5. Exibindo o gráfico
fig.show()
```

## Top 7 Média de Availiações Feitas Por País

```python
# 1. Agrupando por país e calculando a média de avaliações (votos)
media_avaliacoes_por_pais = (
    df.groupby("country")["votes"]
    .mean() # Calcula a média
    .round(2) # Arredonda para 2 casas decimais
    .reset_index(name="media_avaliacoes")
    .sort_values(by="media_avaliacoes", ascending=False)
)

# 2. Pegando o Top 7
top_7_media_avaliacoes = media_avaliacoes_por_pais.head(7)

# 3. Criando o gráfico
fig = px.bar(
    top_7_media_avaliacoes, 
    x="country", 
    y="media_avaliacoes", 
    title="Top 7 Média de Avaliações Feitas por País",
    labels={
        "country": "País", 
        "media_avaliacoes": "Média de Avaliações (Votos)"
    },
    text_auto=True,
    color="country", 
    color_discrete_sequence=px.colors.qualitative.Safe 
)

# 4. Ajustando o visual
fig.update_layout(
    xaxis_tickangle=-45, 
    title_x=0.5,
    showlegend=False, # Oculta a legenda pois já temos os países no eixo X
    xaxis_categoryorder='total descending' # Força a ordem da maior para a menor
)

# 5. Exibindo o gráfico
fig.show()
```

## Top 7 Média de Preço de Um Prato Para Duas Pessoas Por País

```python
# 1. Usando a sua referência, agrupando e calculando a média em USD
pais_valor_medio_prato_2_pessoas = (
    df.groupby("country")["average_cost_for_two_cost_in_usd"]
    .mean()
    .round(2) # Arredonda para 2 casas decimais para o gráfico ficar limpo
    .reset_index(name="media_preco_usd") # Dando um nome mais curto para facilitar
    .sort_values(by="media_preco_usd", ascending=False)
)

# 2. Pegando o Top 7
top_7_preco_pais_usd = pais_valor_medio_prato_2_pessoas.head(7)

# 3. Criando o gráfico
fig = px.bar(
    top_7_preco_pais_usd, 
    x="country", 
    y="media_preco_usd", 
    title="Top 7 Média de Preço de um Prato para Duas Pessoas por País (em Dólares - USD)",
    labels={
        "country": "País", 
        "media_preco_usd": "Média de Preço (USD)"
    },
    text_auto=True,
    color="country", 
    color_discrete_sequence=px.colors.qualitative.Safe 
)

# 4. Ajustando o visual
fig.update_layout(
    xaxis_tickangle=-45, 
    title_x=0.5,
    showlegend=False,
    xaxis_categoryorder='total descending' # Força a ordem decrescente
)

# 5. Exibindo o gráfico
fig.show()
```
